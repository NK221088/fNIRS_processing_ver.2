```python
import pandas as pd
import matplotlib.pyplot as plt

beta_data = pd.read_csv(rf"C:\Users\NTres\OneDrive - Danmarks Tekniske Universitet\Bachelor_projekt\Results\glm_betas_summary.csv")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

# HC group
HC_data = beta_data[beta_data['Group'] == 'HC']
for subject in HC_data['ID'].unique():
    subject_data = HC_data[HC_data['ID'] == subject].sort_values('Condition')
    ax1.plot(subject_data['Condition'], subject_data['theta'], 'o-', alpha=0.5, color='blue')
ax1.set_xlabel('Condition')
ax1.set_ylabel('Theta Values')
ax1.set_title('HC Group')

# Patient group
patient_data = beta_data[beta_data['Group'] == 'Patient']
for subject in patient_data['ID'].unique():
    subject_data = patient_data[patient_data['ID'] == subject].sort_values('Condition')
    ax2.plot(subject_data['Condition'], subject_data['theta'], 'o-', alpha=0.5, color='red')
ax2.set_xlabel('Condition')
ax2.set_title('Patient Group')

plt.tight_layout()
plt.savefig(rf"C:\Users\NTres\OneDrive - Danmarks Tekniske Universitet\Bachelor_projekt\Results\theta_line_plots.png")
plt.close()
```
